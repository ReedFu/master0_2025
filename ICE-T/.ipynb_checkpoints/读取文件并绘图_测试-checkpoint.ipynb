```python
import pandas as pd

data = []

columns = ['STRT_DATE/TIME','STOP_DATE/TIME','LAT','LON','ALT','CVI','CVIF','AER','AER_S','IN_RAW','IN_COR','IN_COR_S','IN_SIG','T','T_S','W_SAT','W_SAT_S','P','P_S']

df = pd.read_csv(r"E:\data\ICE-T\INP\CSU_cfdc_rf01_v1.txt", names=columns, skiprows=23, sep='\s+')

print(df)
```

```python
import numpy as np
import matplotlib.pyplot as plt

#plt.rcParams['font.sans-serif']=['SimHei']#中文正常显示，但会导致英文符号无法显示

x = df['T']
y = df['IN_COR']

fig, ax = plt.subplots(dpi=500)

ax.scatter(x,y)
ax.set_yscale('log')

ax.set_xlabel('Temperature(℃)')
ax.set_ylabel('INP concentration(L^-3)')

ax.tick_params(which='both', top=True, bottom=True, left=True, right=True, direction='in')

#ax.set_xlim(-30,-12)
#ax.set_ylim(1,10**6)
```
