# 污染物时间序列

## 1479A站点

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python
df = pd.read_csv("D:\\Coding\\Data\\Lanzhou_pollution\\Xiaolei_Wang\\1479A_data.csv")
df.head()
```

```python
df = df[(df['Date']>='2024-09-17') & (df['Date']<='2025-10-1')]
```

### AQI

```python
fig, ax = plt.subplots(figsize=(15, 7))

ax.scatter(pd.to_datetime(df['Date']), df['AQI'], s=10, label='AQI')

ax.xaxis.set_minor_locator(mdates.MonthLocator(interval=1))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y/%m/%d"))
fig.autofmt_xdate()   # 自动旋转日期标签
ax.set_ylabel('AQI')
#ax.set_yscale('log')
ax.grid(alpha=0.3, linestyle='--')
ax.legend()
```

### PM2.5

```python
fig, ax1 = plt.subplots(figsize=(15, 7))
y1 = 'PM2.5'
y2 = 'PM2.5_24h'
ax1.scatter(pd.to_datetime(df['Date']), df[y1], s=10, label=y1, color='C0')
ax2 = ax1.twinx()
ax2.scatter(pd.to_datetime(df['Date']), df[y2], s=10, label=y2, color='C1')
    
ax1.xaxis.set_minor_locator(mdates.MonthLocator(interval=1))
ax1.xaxis.set_major_formatter(mdates.DateFormatter("%Y/%m/%d"))
fig.autofmt_xdate()   # 自动旋转日期标签
ax1.set_ylabel(f'{y1}(μg/m³)', color='C0')
ax2.set_ylabel(f'{y2}(μg/m³)', color='C1', rotation=270, labelpad=15)
#ax1.set_yscale('log')
ax1.grid(alpha=0.3, linestyle='--')
#ax1.legend()
#ax2.legend()
```
