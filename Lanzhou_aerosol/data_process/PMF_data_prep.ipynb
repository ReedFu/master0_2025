```python
# 与INP数据合并, 将文件的时间与INP测量的时间对齐
import pandas as pd

df_inp = pd.read_csv(r"D:\Coding\Data\Lanzhou_aerosol\SMPS+APS\INP+ns(v1.0.2).csv")
df_chemical = pd.read_csv(r"D:\Coding\Data\Lanzhou_chemical\merged_data.csv")
df_inp = df_inp[df_inp['Is_Significant'] == True] # 只保留显著的INP数据

df_inp = df_inp[["Time", "N_INP(#/L)", "n_s", "T_a(degC)", "Season"]]
chemical_units = df_chemical.iloc[0]
df_chemical = df_chemical.iloc[1:]  # 去掉单位行

# 转换时间列为 datetime，并删除无法解析的时间
df_inp["Time"] = pd.to_datetime(df_inp["Time"], errors="raise")
df_chemical["datetime"] = pd.to_datetime(df_chemical["datetime"], errors="raise")

# merge_asof 要求按键排序
df_inp = df_inp.sort_values("Time").reset_index(drop=True)
df_chemical = df_chemical.sort_values("datetime").reset_index(drop=True)

result = pd.merge_asof(
	df_inp,
	df_chemical,
	left_on="Time",
	right_on="datetime",
	direction="nearest",
	tolerance=pd.Timedelta("1h")
)
result.to_csv(r"D:\Coding\Data\Lanzhou_chemical\Corr(INP_vs_element).csv", index=False)# 用于相关性分析的文件, 包含INP数据和化学成分数据
# 删除PMF源解析不需要的列
result = result.drop(columns=["Time", "N_INP(#/L)", "n_s", "T_a(degC)", "Season"])

# 去除重复的 datetime 行
result = result.drop_duplicates(subset=["datetime"])

# 将单位行移到第一行
result = pd.concat([pd.DataFrame([chemical_units], columns=result.columns), result], ignore_index=True)
result.to_csv(r"D:\Coding\Data\Lanzhou_chemical\final_data(240924-250929).csv", index=False)# 仅含有INP数据的行(去重后)
```

```python
# 计算不确定度矩阵(作为PMF的输入文件之一)
import pandas as pd
import numpy as np

# 1. 读取原始浓度数据 (第一列是时间)
df_conc = pd.read_csv(r"D:\Coding\Data\Lanzhou_chemical\final_data(241222-250929).csv", index_col=0)
df_conc = df_conc.iloc[1:] #去除单位行
# 2. 准备一个包含每个物种 MDL 的字典
mdl_dict = { # 单位均为: μg/m³
    'NO3-': 0.02, # 有文献 0.05
    'SO42-': 0.04, # 有文献 0.08
    'NH4+': 0.02, # 有文献 0.05/0.08
    'Cl-': 0.02,

    'Ca2+': 0.04,
    'Na+': 0.04,
    'K+': 0.09,
    'Mg2+': 0.08,

    'SO2': 1.0*2.62, # 常温常压下（25℃，1个标准大气压）, ppb 转 μg/m³ 需要乘以 2.62
    # 'NO': 

    'OC(thermal)': 0.4,
    'OC(optical)': 0.4,
    'EC(thermal)': 0.2,
    'EC(optical)': 0.2,

    'PM2.5': 0.5 # assuming a typical MDL for PM2.5
}

# 所有的金属的MDL都设置为0.002 μg/m³
df_metal = pd.read_csv(r"D:\Coding\Data\Lanzhou_chemical\Metal\metal_data.csv")

for col in df_metal.columns:
    if col != "datetime":
        mdl_dict[col] = 0.002

error_fraction = 0.10 # 假定误差比例为10%

def calculate_uncertainty(val, mdl):
    if isinstance(val, str):
        try:
            val = float(val)       
        except ValueError:
            raise ValueError(f"Unexpected string value '{val}' encountered.")
        
    if pd.isna(val) or val == -999:
        return -999  # 缺失值保留标记
    elif val <= mdl:
        return (5/6) * mdl
    else:
        return np.sqrt((error_fraction * val)**2 + (0.5 * mdl)**2)

# 3. 计算不确定度矩阵
df_unc = df_conc.copy()
for col in df_conc.columns:
    mdl = mdl_dict[col]
    df_unc[col] = df_conc[col].apply(lambda x: calculate_uncertainty(x, mdl))

# 4. 导出为 CSV 文件供 PMF 软件使用
df_unc.to_csv(r"D:\Coding\Data\Lanzhou_chemical\Uncertainty_all.csv")
```
