# 绘制兰州市降水时间序列

Precipitation time series from 2024.02.01 to 2025.11.11 at Lanzhou, Gansu province, China

Time resolution: 3h

data source: rp5.ru

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
```

```python
# 读取降水数据
data = pd.read_csv(r"D:\Coding\Data\Lanzhou_precipitation\rp5.ru\52983.01.02.2024.11.11.2025.1.0.0.en.utf8.00000000.csv"
    , skiprows=6
    , sep=";"
    , index_col=False
)
data.head()
```

```python
# 绘制2024年2月到2025年10月的降水时间序列

## 设定时间区间
#date = pd.to_datetime(data['Local time in Lanzhou'], format='%d.%m.%Y %H:%M')
#data_jul_2024 = data[(date >= '2024-07-01') & (date < '2024-08-01')]# 更换时间, 记得修改

x = pd.to_datetime(data['Local time in Lanzhou'], format='%d.%m.%Y %H:%M')
y = data['RRR'].replace(['No precipitation', 'Trace of precipitation'], [0, 0]).astype(float)

fig = plt.figure(dpi=1000)
ax = fig.add_subplot(111)

ax.bar(x,y,color='tab:blue'
       #, edgecolor='k', width=0.5
       )

## 每隔1个月设置一个主刻度
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=1))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d'))
plt.gcf().autofmt_xdate()# 自动旋转日期标记
ax.set_xlabel('Date')
ax.set_ylabel('Precipitation (mm/3h)')
ax.tick_params(which='both', top=True, bottom=True, left=True, right=True, direction='in')
ax.set_title('Lanzhou Precipitation Time Series')
plt.show()

#保存图片
#plt.savefig('Lanzhou_Precipitation_24to25.png', dpi=1000)
```
